```python
'''1. Count Characters and Remove Spaces 
Ask the user to enter a sentence. 
Print the same sentence without spaces. 
Print: Total number of characters (without using Len())  '''

sent=input("Enter a sentence:")
space=" "
new_sent=""
count_ch=0
for s in sent:
    if s!= space:
        new_sent+=s
        count_ch+=1
print(f"sentence without space is :{new_sent}")
print(f"Total number of chracters is :{count_ch}")
```

```python
'''5. Longest Word 
Ask the user to enter a sentence. 
Print the longest word.'''

sentence = input("Enter a sentence: ")
words = sentence.split()
longest = ""
for w in words:
    if len(w) > len(longest):
        longest = w
print("Longest word:", longest)
```

```python
'''6. Count Upper and Lower Case 
Ask the user to enter a string. Print: 
• Number of uppercase letters  
• Number of lowercase letters  
Example: 
HeLLo → Upper = 3, Lower = 2'''

string=input("Enter a string:")
count_l=0
count_u=0
for s in string:
    if s ==s.lower():
        count_l+=1
    elif s==s.upper():
        count_u+=1
    else:
        print("please Enter a string")
print(f"Number of lowercase letters:{count_l}")
print(f"Number of uppercase letters:{count_u}")
```

```python
'''9. Prime Number Check 
Ask the user to enter a number. 
Print whether it is Prime or Not.'''

num = int(input("Enter a number: "))
count = 0
for i in range(1, num + 1):
    if num % i == 0:
        count += 1
if count == 2:
    print(f"{num}:Prime")
else:
    print(f"{num}:Not Prime")
```

```python
'''Task 3: Reverse a List using while 
Write a Python program to reverse a list using while loop only (no slicing or built-in functions).'''
mylist=[1,2,3,4,5,6,7]
rev=[]
i=len(mylist)-1
while i >=0:
    rev.append(mylist[i])
    i-=1
print(rev)

lst = [1, 2, 3]
rev = []
i = 0
while i < len(lst):
    rev.insert(0, lst[i])  
    i += 1
print(rev)
```

```python
'''Dictionary: 
Task 1: 
Given a dictionary of students and grades, find the student with the highest grade without using built-in 
functions'''
students={"Ahmed":300,
         "Menna":400,
         "sara":399,}
highest=-1
top_stu=""
for name in students:
    grade=students[name]
    if grade> highest:
        highest=grade
        top_stu=name
print("Top student:", top_stu)
print("Highest grade:", highest)
```

```python
'''Task 3: 
Write a Python program that takes a list of integers and creates a dictionary that stores how many times each 
number appears using while loop only (no count(),no in).'''
nums = [1, 2, 3, 2, 1, 3, 3, 4]
result = {}
i = 0
while i < len(nums):
    current = nums[i]
    # لو المفتاح مش موجود هيكون قيمته 0 تلقائيًا
    if result.get(current) == None:
        result[current] = 1
    else:
        result[current] += 1
    i += 1
print(result)
```
