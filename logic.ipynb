```python
def generate_negation_table():
    propositions=[True,False]
    print("p \t | ¬p")
    print("-"*20)
    for p in propositions:
        negation=not p 
        print(f"{p}\t | {negation}")
generate_negation_table()
```

```python
def draw_conjunction_table():
    values=[True,False]
    print("Truth table for conjunction (p^q)")
    print("p\t|q\t|p and q")
    print("-"*30)
    for p in values:
        for q in values:
            result= p and q
            print(f"{p}\t|{q}\t|{result}")
draw_conjunction_table()
```

```python
def draw_or_table():
    values=[True,False]
    print("Truth table for disjunction(pvq)")
    print("p\t|q\t|p or q")
    print("-"*30)
    for p in values:
        for q in values:
            result=p or q 
            print(f"{p}\t|{q}\t|{result}")
draw_or_table()
```

```python
def draw_XOR_table():
    values=[True,False]
    print("p\t|q\t|p xor q")
    print("-"*30)
    for p in values:
        for q in values:
            result=p^q
            print(f"{p}\t|{q}\t|{result}")
draw_XOR_table()
```

```python
def draw_implication_table():
    values=[True,False]
    print("p\t|q\t|p -> q")
    print("-"*30)
    for p in values:
        for q in values:
            result=not(p)or q 
            print(f"{p}\t|{q}\t|{result}")
draw_implication_table()
print("_"*30)
print("p\t|q\t|p -> q")
print("-"*30)
values=[True,False]
for p in values:
    for q in values:
        if p:
            result=q
        else:
            result=True
        print(f"{p}\t|{q}\t|{result}")
```

```python
def draw_biconditional_table():
    values=[True,False]
    print("p\t|q\t|p <-> q")
    print("_"*30)
    for p in values:
        for q in values:
            result=(p==q)
            print(f"{p}\t|{q}\t|{result}")
draw_biconditional_table()
```

```python
def formula():
    print(f"p\t|q\t|(pv!q)\t|(p^q)\t|Result")
    print("-" *50)
    for p in [True, False]:
        for q in [True, False]:
            part1 = p or (not q)
            part2 = p and q
            final_result = (not part1) or part2
            print(f"{p}\t|{q}\t|{part1}\t|{part2}\t|{final_result}")
formula()
```
