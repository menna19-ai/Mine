```python
'''Task 1: Word Frequency Analyzer
Requirements:
Read text data from a file
Count the frequency of each word using a dictionary
Ignore case sensitivity
Ignore punctuation marks
Display the top 3 most frequent words
Handle file-related exceptions properly'''
filename= input("Enter a file name:")
freq={}
try:
    with open(filename,"r") as file:
        content=file.read().lower()
        for ch in ".,!?;:'\"":
            content = content.replace(ch, "")
        words=content.split()
        for word in words:
          freq[word]=freq.get(word,0)+1
        for _ in range(3):
            most_word = ""
            most_count = -1
            for word in freq:
               if freq[word] > most_count: ###
                   most_count = freq[word]
                   most_word = word
            if most_word == "":
                break 
    print(most_word, ":", most_count)
    del freq[most_word]
except FileNotFoundError:
    print("file not found Enter the right name!")
```

```python
'''Task 2: File-Based Login System
Requirements:
Read user credentials from a file
Implement a login function that checks username and password
Display a success or failure message
Handle missing file errors
Prevent the program from crashing on invalid input'''
filename=input("Enter a file name:")
def checking(username,password):
    special = "@#$&*"
    if not (any(c.isalpha() for c in username) and any(c.isdigit() for c in username)): ###
        print("username must contain letters and digits")
        return False
    if (len(password) <= 8 or
        not any(c.isalpha() for c in password) or 
        not any(c.isdigit() for c in password) or
        not any(c in special for c in password)):
        print("password must contain letters, digits, special char, and > 8 chars")
        return False
    return True
try:
    with open(filename,"r") as file:
        username = input("Username: ")
        password = input("Password: ")
        for line in file:
            file_user, file_pass = line.strip().split(",")
            if checking(username,password):
                if username == file_user and password == file_pass:
                    print("Login success")
                else:
                    print("login Fail!")
except FileNotFoundError:
    print("file not found,please Enter the right name!")
```

```python
''' Task 3: Student Records System
Requirements:
Store student data using a nested dictionary
Create a function without parameters to load data from a file
Create a function with parameters to add a new student
Calculate the average grade for each student
Save updated data back to the file
Handle file-related exceptions'''
students={}
def load_data():
    filename=input("Enter file name:")
    try:
        with open(filename,"r")as file:
            for line in file:
                name,grade=line.strip().split(",")
                grade=float(grade)
                students[name]={"grades":[grade],"average":grade}
    except FileNotFoundError:
        print("File not found!Enter the right name")
def add_student(studentname,grade):
    try:
        grade=float(grade)
        students[studentname] = {
            "grades": [grade],
            "average": grade}
    except ValueError:
        print("grade should be digit")
def main():
    load_data()
    while True:
        print("1.add student\n2.load data\n3.exit")
        choice=input("Enter your option:")
        if choice=="1":
            name = input("Student name: ")
            grade = input("Grade: ")
            add_student(name, grade)
        elif choice=="2":
            load_data()
        elif choice=="3":
            break
        else:
            print("Enter the right choice")
```

```python
''' Task 4: Inventory Management System
Requirements:
Store products using a nested dictionary
Create a function to add a new product
Create a function to update product price
Create a function to reduce quantity after a sale
Calculate the total inventory value
Handle errors such as missing products or invalid operations
'''
products={}
def add_new_product(productname,price,quantity):
    try:
        price=float(price)
        quantity=int(quantity)
        if price <= 0 or quantity <= 0:
            print("price and quantity must be positive")
            return
        products[productname]={"price":price,"quantity":quantity}
    except ValueError as e:
        print(e)
        
def update_product_price(product_name,new_price):
    if product_name not in products:
        print("product not found!")
        return
    try:
        new_price = float(new_price)
        if new_price <= 0: 
            print("price must be positive")
            return
        products[product_name]["price"] = new_price
    except ValueError:
        print("invalid price")
        
def reduce_quantity(sold_product,sold_quantity):
    if sold_product not in products:
        print("product not found!")
        return
    sold_quantity = int(sold_quantity)
    if sold_quantity <= 0:
        print("invalid quantity")
        return
    if products[sold_product]["quantity"] < sold_quantity:
        print("not enough stock")
        return
    products[sold_product]["quantity"] -= sold_quantity  
    print(products[sold_product]["quantity"])
    
def total_inventory_value():
    total = 0
    for product in products.values():
        total += product["price"] * product["quantity"] ##
    print("Total value =", total)

def main():
    while True:
        print("1.add new product\n2.update product price\n3.reduce quantity\n4.total inventory value\n5.Exit\n")
        choice=input("Enter your choice:")
        if choice=="1":
            product_name=input("Enter product name:").lower()
            price=input("Enter product price:")
            quantity=input("Enter product quantity:")
            add_new_product(product_name,price,quantity)
        elif choice =="2":
            product_name=input("Enter product name:").lower()
            new_price=input("Enter the new price:")
            update_product_price(product_name,new_price)
        elif choice=="3":
            sold_product=input("Enter sold product:").lower()
            sold_quantity=input("Enter sold quantity:")
            reduce_quantity(sold_product,sold_quantity)
        elif choice=="4":
            total_inventory_value()
        elif choice=="5":
            break
        else:
            print("Please Enter the right number!")
main()
```

```python
'''Task 5: Nested Dictionary Analyzer
Requirements:
Work with a nested dictionary representing structured data
Create a function without parameters to display all data
Create a function with parameters to find the highest value
Calculate average values for grouped data
Add new entries dynamically
Handle errors when accessing missing keys'''
data={"menna":{"grade":400},"sara":{"grade":399}}
def show_all_data():
    if len(data)==0:
        print("there is no data!")
    for key,value in data.items():
        print(f"{key}:{value}\n")
        
def highest_value(data):
    highest=-1
    top_student=""
    for key, value in data.items():
        grade = value["grade"] ##
        if grade > highest:
            highest = grade
            top_student = key
    print("Top student:", top_student)
    print("Highest grade:", highest)

def average(data):
    for name, value in data.items():
        grade = value["grade"]  
        print(f"{name} average: {grade}")

def add_new_values(name,grade):
    try:
        grade=float(grade)
        data[name]={"grade":grade}
    except ValueError:
        print("Invalid operation!")   

def main():
    while True:
        print("1.add new value\n2.highest value\n3.average\n4.show all data\n5.Exit\n")
        choice=input("Enter your choice:")
        if choice=="1":
            name=input("Enter student name:").lower()
            grade=input("Enter student grade:")
            add_new_values(name,grade)
        elif choice=="2":
            highest_value(data)
        elif choice=="3":
            average(data)
        elif choice =="4":
            show_all_data()
        elif choice=="5":
            break
        else:
            print("Enter the right number!")
main()
```
